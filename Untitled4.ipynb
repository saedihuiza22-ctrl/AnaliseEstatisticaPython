```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import drive
import sqlite3
```

```python
print("="*50)
print("📊 ANÁLISE DE DADOS - NUBANK 2027")
print("="*50)
```

```python
print("\n Gerando dados de exemplo...")
quantidade = 200
np.random.seed(42)

tabela = pd.DataFrame({
    'idade': np.random.randint(18, 65, quantidade),
    'salario': np.random.randint(2000, 15000, quantidade),
    'experiencia': np.random.randint(0, 40, quantidade),
    'avaliacao': np.random.uniform(1, 5, quantidade).round(1),
    'departamento': np.random.choice(['TI', 'RH', 'Vendas', 'Marketing', 'Financeiro'], quantidade),
    'cidade': np.random.choice(['SP', 'RJ', 'BH', 'Curitiba', 'POA'], quantidade)
})

print(f" {quantidade} registros criados!")
print("\n Primeiras linhas:")
```

```python
print(tabela.head())
```

```python
print("\n" + "="*50)
print(" ANÁLISE ESTATÍSTICA")
print("="*50)

colunas_numericas = tabela.select_dtypes(include=[np.number]).columns.tolist()
print(f"\n Colunas numéricas: {colunas_numericas}")

print("\n Estatísticas Descritivas:")
print(tabela[colunas_numericas].describe())
```

```python
print("\n" + "="*50)
print(" PROCURANDO OUTLIERS")
print("="*50)

if colunas_numericas:
    coluna = colunas_numericas[0]
    print(f"\n Analisando: {coluna}")

    Q1 = tabela[coluna].quantile(0.25)
    Q3 = tabela[coluna].quantile(0.75)
    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    outliers = tabela[
        (tabela[coluna] < limite_inferior) |
        (tabela[coluna] > limite_superior)
    ]

    print(f"Outliers encontrados: {len(outliers)}")
```

```python
print("\n" + "="*50)
print(" CORRELAÇÃO")
print("="*50)

if len(colunas_numericas) >= 2:
    print("\nMatriz de correlação:")
    print(tabela[colunas_numericas].corr())
```

```python
print("\n" + "="*50)
print(" GRÁFICOS")
print("="*50)

if colunas_numericas:
    coluna = colunas_numericas[0]

    print(f"\n Histograma de {coluna}")
    plt.figure(figsize=(10, 5))
    tabela[coluna].hist(bins=20, alpha=0.7, color='skyblue', edgecolor='black')
    tabela[coluna].plot.kde(color='red', linewidth=2)
    plt.title(f'Distribuição - {coluna}', fontsize=14)
    plt.xlabel(coluna)
    plt.ylabel('Frequência')
    plt.grid(True, alpha=0.3)
    plt.show()
```

```python
    print(f"\n Boxplot de {coluna}")
    plt.figure(figsize=(8, 5))
    tabela.boxplot(column=[coluna], patch_artist=True)
    plt.title(f'Boxplot - {coluna}', fontsize=14)
    plt.grid(True, alpha=0.3)
    plt.show()
```

```python
if len(colunas_numericas) >= 2:
    print("\n Heatmap de correlação")
    plt.figure(figsize=(10, 8))
    correlacao = tabela[colunas_numericas].corr()
    sns.heatmap(correlacao, annot=True, fmt='.2f', cmap='coolwarm',
                center=0, square=True, linewidths=0.5)
    plt.title('Matriz de Correlação', fontsize=14)
    plt.tight_layout()
    plt.show()
```

```python
print("\n" + "="*50)
print(" BANCO DE DADOS SQL")
print("="*50)

conexao = sqlite3.connect('analise.db')
tabela.to_sql('dados', conexao, if_exists='replace', index=False)
print(" Dados salvos no SQL")
```

```python

print("\n  Consulta SQL - Média por departamento:")
if 'departamento' in tabela.columns and colunas_numericas:
    query = f"""
    SELECT departamento,
           AVG({colunas_numericas[0]}) as media,
           COUNT(*) as quantidade
    FROM dados
    GROUP BY departamento
    ORDER BY media DESC
    """
    resultado = pd.read_sql_query(query, conexao)
    print(resultado)

conexao.close()


print("\n" + "="*50)
print(" ANÁLISE CONCLUÍDA COM SUCESSO!")
print("="*50)

print("\n RESUMO:")
print(f"- Total de registros: {len(tabela)}")
print(f"- Colunas numéricas: {len(colunas_numericas)}")

```
